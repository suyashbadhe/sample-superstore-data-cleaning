```python
discount_analysis = df.groupby("Discount").agg(
    Total_Orders=("Order ID", "count"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Average_Sales=("Sales", "mean"),
    Average_Profit=("Profit", "mean")
)

display(discount_analysis
)
```

```python
loss_analysis = df[df["Profit"] < 0].groupby("Category").agg(
    Loss_Records=("Profit", "count"),
    Total_Loss=("Profit", "sum")
)

display(loss_analysis)
```

```python
product_loss = df[df["Profit"] < 0].groupby("Product Name").agg(
    Loss_Records=("Profit", "count"),
    Total_Loss=("Profit", "sum")
).sort_values("Total_Loss")

display(product_loss.head(10))
```

```python
print("Sales <= 0:", (df["Sales"] <= 0).sum())
print("Quantity <= 0:", (df["Quantity"] <= 0).sum())
print("Discount < 0:", (df["Discount"] < 0).sum())
print("Discount > 1:", (df["Discount"] > 1).sum())
```

```python
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    empty_count = df[col].apply(
        lambda x: isinstance(x, str) and x.strip() == ""
    ).sum()
    print(col, ":", empty_count)
```
